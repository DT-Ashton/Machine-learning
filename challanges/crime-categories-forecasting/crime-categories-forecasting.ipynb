```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
```

```python
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')

train.info()
```

```python
train.head(10)
```

```python
train = train.drop(columns=['Location', 'Cross_Street', 'Date_Reported', 'Date_Occurred', 'Area_Name', 'Premise_Description', 'Weapon_Description', 'Status_Description'])
test = test.drop(columns=['Location', 'Cross_Street', 'Date_Reported', 'Date_Occurred', 'Area_Name', 'Premise_Description', 'Weapon_Description', 'Status_Description'])
```

```python
# Label encoding
label_encode = LabelEncoder()
train['Victim_Sex'] = label_encode.fit_transform(train['Victim_Sex'])
train['Victim_Descent'] = label_encode.fit_transform(train['Victim_Descent'])
train['Status'] = label_encode.fit_transform(train['Status'])

test['Victim_Sex'] = label_encode.fit_transform(test['Victim_Sex'])
test['Victim_Descent'] = label_encode.fit_transform(test['Victim_Descent'])
test['Status'] = label_encode.fit_transform(test['Status'])

y_train = train['Crime_Category']
```

```python
train['Weapon_Used_Code'] = train['Weapon_Used_Code'].fillna(0)
train.info()
```

```python
test['Weapon_Used_Code'] = test['Weapon_Used_Code'].fillna(0)
test.info()
```

```python
train['Modus_Operandi'] = train['Modus_Operandi'].astype(str)
train['Modus_Operandi'] = train['Modus_Operandi'].fillna('').apply(lambda x: x.split())

test['Modus_Operandi'] = test['Modus_Operandi'].astype(str)
test['Modus_Operandi'] = test['Modus_Operandi'].fillna('').apply(lambda x: x.split())

from sklearn.preprocessing import MultiLabelBinarizer
mlb = MultiLabelBinarizer()

modus_encoded = mlb.fit_transform(train['Modus_Operandi'])
modus_encoded_train = pd.DataFrame(modus_encoded, columns=mlb.classes_, index=train.index)
train = train.join(modus_encoded_train).drop(columns=['Modus_Operandi'])

modus_encoded = mlb.fit_transform(test['Modus_Operandi'])
modus_encoded_test = pd.DataFrame(modus_encoded, columns=mlb.classes_, index=test.index)
test = test.join(modus_encoded_test).drop(columns=['Modus_Operandi'])

train
```

```python
for col in train.columns:
    if col not in test.columns:
        test[col] = 0
test = test[train.columns]
test
```

```python
# Modeling
x = train.drop(columns='Crime_Category')
y = train['Crime_Category']
model = (RandomForestClassifier(random_state=42))  
model.fit(x,y)
prediction = model.predict(test.drop(columns='Crime_Category'))
```

```python
submission = pd.DataFrame(columns=["ID", "Crime_Category"])
submission['Crime_Category'] = prediction
submission['ID'] = submission['Crime_Category'].index + 1
submission.to_csv('crime-categories-forecasting.csv', index=False)
submission
```
