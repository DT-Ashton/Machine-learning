```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
```

```python
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
x_train = train.drop('label', axis='columns')
y_train = train['label']
x_train = x_train.to_numpy()
y_train = y_train.to_numpy()

for row in x_train:
    row = row.reshape(28,28)

print(x_train.shape)
print(y_train.shape)
```

```python
x_test = test.to_numpy()
for row in x_test:
    row = row.reshape(28,28)
```

```python
knn = KNeighborsClassifier(n_neighbors=9) 
knn.fit(x_train,y_train)
predict = knn.predict(x_test)

print(x_test.shape)
print(predict.shape)
```

```python
result = pd.DataFrame(range(1,28001))
result = result.rename(columns={0: 'ImageId'})
result['Label'] = predict
result
```
